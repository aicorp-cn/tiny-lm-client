# 提示词工程工具

本 notebook 用于生成优化的提示词。

```python
client = TinyLMClient(
    base_url="http://localhost:11434/v1",
    api_key="your-api-key-here",
    max_retries=3
)

print("客户端初始化完成")
```

```python
models = await client.models_list()
for model in models:
    print(f"Model: {model.id}, Owner: {model.owned_by}")
```

```python
from tiny_lm_client import ChatCompletionRequest, Message
messages = [
    {"role": "user", "content": "Hello"}
]

request = ChatCompletionRequest(
    model="deepseek-r1:1.5b",
    messages=[Message(**msg) for msg in messages],
    temperature=0.7,
    max_tokens=4000
)

response = await client.chat_completion(request)
print("Response: ",response.choices[0].message.content)
```

```python
async def gen(user_input: str) -> str:
    from tiny_lm_client import ChatCompletionRequest, Message
    
    sys_prompt = Path("./prompts/sys_r1_craft_prompt_framework.md").read_text(encoding="UTF-8")
    
    messages = [
        {"role": "system", "content": sys_prompt},
        {"role": "user", "content": user_input}
    ]
    
    request = ChatCompletionRequest(
        model="gpt-4",
        messages=[Message(**msg) for msg in messages],
        temperature=0.7,
        max_tokens=4000
    )
    
    response = await client.chat_completion(request)
    return response.choices[0].message.content
```

```python
result = await gen("写一篇关于苹果 M5 MacBook Max 的评测试博文")
print(result)
```
